### make NN

```python
import tensorflow as tf
import numpy as np

# Generate random training data (1000 samples, 10 features)
X_train = np.random.rand(1000, 10)
y_train = np.random.randint(0, 2, size=(1000,))  # Binary classification (0 or 1)

# Define a simple neural network model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(32, activation='relu', input_shape=(10,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')  # Binary output (0 or 1)
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model (quick test)
model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=1)
```
